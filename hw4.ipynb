Nhập vào số nguyên dương n, tính tổng các chữ số của n

```python
n = int(input("Nhập số: "))
tong = 0

while n > 0:
    chu_so_cuoi = n % 10   
    tong = tong + chu_so_cuoi  
    n = n // 10     
    
print("Tổng là:", tong)
```

 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n=int(input())
a=0
b=0
while n>0:
    so_dang_xet=n%10
    if so_dang_xet%2==0:
        a=a+1
    else:
        b=b+1
        n=n//10
print(a)
print(b)


```

Nhập n
Cho S(k) = 1 + 2 + 3 + … + k
Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 


```python
n=int(input())
S=0
k=0
while S<n:
    k=k+1
    S=S+k
k=k-1
print(k)
```

Nhập vào A, tìm n nhỏ nhất sao cho
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 

```python
A=int(input())
n=0
s=0
while s<A:
    n=n+1
    s=s+1/n
print(n)
```

[Factorial]
Viết chương trình tính giai thừa của một số nguyên n cho trước (n!=1×2×…×n)
với 0 < n < 100

```python
n=int(input())
s=0
for i in range(1,n+1):
    s =s + i * (i - 1)
print(s)
```
