Viết hàm hoán đổi hai số nguyên. def swap(a, b) -> Tuple[int, int]: 


```python
a,b=map(int,input().split())
def swap(a, b) -> Tuple[int, int]:
    a,b=b,a
    return a,b
print (swap(a,b))
       
```
